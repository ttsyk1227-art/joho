### 課題1（基本）
身長（cm）と体重（kg）を入力させ、BMI（体重 ÷ 身長(m)²）を表示しなさい。身長は小数も入力できるようにすること。

**実行例**
```
身長(cm): 165
体重(kg): 55
BMIは20.2です。
```

````python
height = float(input("身長(cm): "))
weight = float(input("体重(kg): "))
bmi = weight / (height / 100) ** 2
print(f"BMIは{round(bmi, 1)}です。")
````

### 課題2（基本）
秒数を入力させ、それが何分何秒かを表示しなさい。

**実行例**
```
秒数: 200
200秒は3分20秒です。
```

````python
sec = int(input("秒数: "))
print(f"{sec}秒は{sec // 60}分{sec % 60}秒です。")
````

### 課題3（応用）
商品の値段と個数、支払った金額を入力させ、おつりを表示しなさい。

**実行例**
```
値段: 120
個数: 3
支払い: 1000
合計は360円、おつりは640円です。
```

````python
price = int(input("値段: "))
count = int(input("個数: "))
pay = int(input("支払い: "))
total = price * count
print(f"合計は{total}円、おつりは{pay - total}円です。")
````
