### 課題1（基本）
正の整数が入力されるまで入力を繰り返し、正の整数が入力されたら「OK」と表示して終了しなさい。

**実行例**
```
正の整数: -3
もう一度入力してください
正の整数: 0
もう一度入力してください
正の整数: 5
OK
```

````python
while True:
    n = int(input("正の整数: "))
    if n > 0:
        print("OK")
        break
    print("もう一度入力してください")
````

### 課題2（基本）
数を次々に入力させ、0 が入力されたら、それまでに入力された数の合計を表示して終了しなさい。

**実行例**
```
数: 5
数: 8
数: 2
数: 0
合計は15です。
```

````python
total = 0
while True:
    n = int(input("数: "))
    if n == 0:
        break
    total += n
print(f"合計は{total}です。")
````

### 課題3（応用）
数当てゲームを作りなさい。答えを 37 とし、入力された数が答えより大きければ「もっと小さい」、小さければ「もっと大きい」と表示し、当たったら何回目で当たったかを表示して終了する。

**実行例**
```
予想: 50
もっと小さい
予想: 30
もっと大きい
予想: 37
正解！3回目で当たりました。
```

````python
answer = 37
count = 0
while True:
    guess = int(input("予想: "))
    count += 1
    if guess > answer:
        print("もっと小さい")
    elif guess < answer:
        print("もっと大きい")
    else:
        print(f"正解！{count}回目で当たりました。")
        break
````
