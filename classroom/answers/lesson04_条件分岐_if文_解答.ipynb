### 課題1（基本）
整数を入力させ、偶数なら「偶数」、奇数なら「奇数」と表示しなさい。

**実行例**
```
整数: 7
奇数
```

````python
n = int(input("整数: "))
if n % 2 == 0:
    print("偶数")
else:
    print("奇数")
````

### 課題2（基本）
2 つの整数を入力させ、大きい方を表示しなさい。等しい場合は「同じです」と表示する。

**実行例**
```
1つ目: 12
2つ目: 30
大きい方は30です。
```

````python
a = int(input("1つ目: "))
b = int(input("2つ目: "))
if a > b:
    print(f"大きい方は{a}です。")
elif a < b:
    print(f"大きい方は{b}です。")
else:
    print("同じです")
````

### 課題3（応用）
西暦を入力させ、うるう年かどうかを判定しなさい。
（4 で割り切れる年はうるう年。ただし 100 で割り切れる年は平年。ただし 400 で割り切れる年はうるう年）

**実行例**
```
西暦: 2024
2024年はうるう年です。
```

````python
year = int(input("西暦: "))
if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print(f"{year}年はうるう年です。")
else:
    print(f"{year}年は平年です。")
````
