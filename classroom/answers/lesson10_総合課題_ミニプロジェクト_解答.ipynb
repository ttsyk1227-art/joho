### 課題（総合）
次のうち 1 つを選び（または自分で考えたテーマで）プログラムを作りなさい。

- **A. 成績処理**：5 人分の点数を入力させ、平均・最高・最低と、各人の評価（A/B/C）を表示する
- **B. 計算ドリル**：ランダムな足し算の問題を 5 問出題し、正解数を表示する
- **C. 自由テーマ**：自分で考えたプログラム（何を作るかを最初にテキストセルで説明すること）

**評価のポイント**：①正しく動く ②学んだ内容を 3 つ以上使っている ③コメントや変数名がわかりやすい ④工夫がある

**実行例**
```
（選んだテーマによる）
```

````python
# --- A. 成績処理 の解答例 ---
def grade(score):
    if score >= 80:
        return "A"
    elif score >= 60:
        return "B"
    else:
        return "C"

scores = []
for i in range(5):
    scores.append(int(input(f"{i + 1}人目の点数: ")))

print(f"平均: {sum(scores) / len(scores)}  最高: {max(scores)}  最低: {min(scores)}")
for i in range(5):
    print(f"{i + 1}人目: {scores[i]}点 {grade(scores[i])}")

# --- B. 計算ドリル の解答例 ---
import random

correct = 0
for i in range(5):
    a = random.randint(1, 20)
    b = random.randint(1, 20)
    ans = int(input(f"第{i + 1}問 {a} + {b} = "))
    if ans == a + b:
        print("正解！")
        correct += 1
    else:
        print(f"ざんねん。正解は{a + b}")
print(f"5問中{correct}問正解でした。")
````
