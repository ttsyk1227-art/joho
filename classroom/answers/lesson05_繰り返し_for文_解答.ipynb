### 課題1（基本）
整数 n を入力させ、n の段の九九を表示しなさい。

**実行例**
```
何の段?: 7
7 × 1 = 7
7 × 2 = 14
…
7 × 9 = 63
```

````python
n = int(input("何の段?: "))
for i in range(1, 10):
    print(f"{n} × {i} = {n * i}")
````

### 課題2（基本）
整数 n を入力させ、1 から n までの合計を表示しなさい。

**実行例**
```
n: 10
1から10までの合計は55です。
```

````python
n = int(input("n: "))
total = 0
for i in range(1, n + 1):
    total += i
print(f"1から{n}までの合計は{total}です。")
````

### 課題3（応用）
二重ループを使って、九九の表（1〜9の段）を表示しなさい。
（ヒント：`print(f"{x:3}", end="")` と書くと 3 桁分の幅で表示できる）

**実行例**
```
  1  2  3  4  5  6  7  8  9
  2  4  6  8 10 12 14 16 18
…
  9 18 27 36 45 54 63 72 81
```

````python
for i in range(1, 10):
    for j in range(1, 10):
        print(f"{i * j:3}", end="")
    print()
````
