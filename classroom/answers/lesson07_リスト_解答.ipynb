### 課題1（基本）
1 週間の最高気温 `[28, 31, 30, 27, 33, 29, 32]` について、平均気温・最高・最低を表示しなさい。

**実行例**
```
平均: 30.0℃
最高: 33℃
最低: 27℃
```

````python
temps = [28, 31, 30, 27, 33, 29, 32]
print(f"平均: {sum(temps) / len(temps)}℃")
print(f"最高: {max(temps)}℃")
print(f"最低: {min(temps)}℃")
````

### 課題2（基本）
空のリストを用意し、5 回入力させた整数を `append` で追加していき、最後にリストと合計を表示しなさい。

**実行例**
```
整数: 3
整数: 8
整数: 1
整数: 6
整数: 2
[3, 8, 1, 6, 2]
合計: 20
```

````python
numbers = []
for i in range(5):
    numbers.append(int(input("整数: ")))
print(numbers)
print("合計:", sum(numbers))
````

### 課題3（応用）
`max()` を使わずに、for 文と if 文だけでリスト `[45, 82, 17, 96, 53]` の最大値を求めなさい。

**実行例**
```
最大値は96です。
```

````python
data = [45, 82, 17, 96, 53]
biggest = data[0]
for d in data:
    if d > biggest:
        biggest = d
print(f"最大値は{biggest}です。")
````
