### 課題1（基本）
おみくじプログラムを作りなさい。「大吉」「中吉」「小吉」「吉」「凶」からランダムに 1 つ選んで表示する。

**実行例**
```
今日の運勢は…中吉！
```

````python
import random

kuji = ["大吉", "中吉", "小吉", "吉", "凶"]
print(f"今日の運勢は…{random.choice(kuji)}！")
````

### 課題2（基本）
直角三角形の 2 辺 a, b を入力させ、斜辺の長さを `math.sqrt()` を使って求めなさい（三平方の定理）。

**実行例**
```
a: 3
b: 4
斜辺の長さは5.0です。
```

````python
import math

a = float(input("a: "))
b = float(input("b: "))
print(f"斜辺の長さは{math.sqrt(a ** 2 + b ** 2)}です。")
````

### 課題3（応用）
1〜100 の整数をランダムに決め、前回の数当てゲーム（第6回 課題3）を完成させなさい。

**実行例**
```
予想: 50
もっと大きい
予想: 75
もっと小さい
…
正解！6回目で当たりました。
```

````python
import random

answer = random.randint(1, 100)
count = 0
while True:
    guess = int(input("予想: "))
    count += 1
    if guess > answer:
        print("もっと小さい")
    elif guess < answer:
        print("もっと大きい")
    else:
        print(f"正解！{count}回目で当たりました。")
        break
````
