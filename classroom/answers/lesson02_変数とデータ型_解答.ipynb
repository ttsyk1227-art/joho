### 課題1（基本）
変数 `price`（税抜価格）を 2980 として、消費税 10% を加えた税込価格を f文字列で表示しなさい。（結果は `int()` で整数にする）

**実行例**
```
税抜2980円の商品の税込価格は3278円です。
```

````python
price = 2980
tax_price = int(price * 1.1)
print(f"税抜{price}円の商品の税込価格は{tax_price}円です。")
````

### 課題2（基本）
国語 78 点、数学 85 点、英語 92 点をそれぞれ変数に入れ、合計点と平均点を表示しなさい。

**実行例**
```
合計点: 255点
平均点: 85.0点
```

````python
kokugo = 78
sugaku = 85
eigo = 92
total = kokugo + sugaku + eigo
print(f"合計点: {total}点")
print(f"平均点: {total / 3}点")
````

### 課題3（応用）
半径 `r = 5` の円の面積と円周を表示しなさい。円周率は 3.14 とする。

**実行例**
```
半径5の円の面積は78.5、円周は31.400000000000002です。
（小数の誤差が出ても正解。round(値, 2) で丸めるとなお良い）
```

````python
r = 5
pi = 3.14
print(f"半径{r}の円の面積は{pi * r ** 2}、円周は{round(2 * pi * r, 2)}です。")
````
