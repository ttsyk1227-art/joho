### 課題2（基本）
1 日は何秒か、1 週間は何秒かを計算して表示しなさい。

**実行例**
```
1日の秒数: 86400
1週間の秒数: 604800
```

````python
print("1日の秒数:", 60 * 60 * 24)
print("1週間の秒数:", 60 * 60 * 24 * 7)
````

### 課題3（応用）
100 分は何時間何分か、`//` と `%` を使って計算し、表示しなさい。

**実行例**
```
100分は 1 時間 40 分
```

````python
print("100分は", 100 // 60, "時間", 100 % 60, "分")
````
