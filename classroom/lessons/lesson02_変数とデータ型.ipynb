# 1. 基本事項

### 変数
**変数** は「値に名前をつけて入れておく箱」です。`=` は「等しい」ではなく **右の値を左の変数に入れる（代入）** という意味です。

````python
price = 150
count = 4
total = price * count
print(total)

count = 10          # 上書きできる
print(price * count)
````

### 変数名のルール
- 英字・数字・`_` が使える（数字から始めてはいけない）
- 大文字と小文字は区別される（`Age` と `age` は別物）
- `print` や `if` などの予約語は使えない
- 意味のわかる名前をつけよう（`a` より `price`）

### データ型
| 型 | 名前 | 例 |
|---|---|---|
| `int` | 整数 | `10`, `-3` |
| `float` | 小数（浮動小数点数） | `3.14`, `0.5` |
| `str` | 文字列 | `"Python"`, `'abc'` |
| `bool` | 真偽値 | `True`, `False` |

`type()` で型を調べられます。`int()` `float()` `str()` で型を変換できます。

````python
print(type(10), type(3.14), type("Python"), type(True))
print("10" + "20")            # 文字列どうしの + はつなげる
print(int("10") + int("20"))  # 整数に変換してから足す
````

### f文字列
文字列の前に `f` をつけると、`{ }` の中に変数や式を埋め込めます。

````python
name = "花子"
age = 16
print(f"{name}さんは{age}歳です。来年は{age + 1}歳です。")
````

# 2. 例題

**【例題】** 縦 6 cm、横 9 cm の長方形について、面積と周の長さを変数を使って計算し、f文字列で表示しなさい。

````python
tate = 6
yoko = 9
menseki = tate * yoko
shu = (tate + yoko) * 2
print(f"縦{tate}cm、横{yoko}cmの長方形")
print(f"面積は{menseki}cm²、周の長さは{shu}cmです。")
````

**【解説】**
- 数値を変数に入れておくと、`tate = 6` の部分を書き換えるだけで別の長方形にも使えます。
- f文字列を使うと `,` で区切るより読みやすい表示になります。
