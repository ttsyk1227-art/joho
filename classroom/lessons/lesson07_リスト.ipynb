# 1. 基本事項

### リストとは
複数の値をまとめて 1 つの変数に入れられるのが **リスト** です。`[ ]` の中に `,` で区切って書きます。
要素には **0 から始まる番号（インデックス）** がついています。`-1` は最後の要素です。

````python
fruits = ["りんご", "バナナ", "みかん"]
print(fruits)
print(fruits[0])    # 先頭
print(fruits[2])
print(fruits[-1])   # 最後
fruits[1] = "ぶどう"  # 書きかえ
print(fruits)
````

### よく使う操作
| 書き方 | 意味 |
|---|---|
| `len(a)` | 要素の数 |
| `a.append(x)` | 最後に x を追加 |
| `sum(a)` / `max(a)` / `min(a)` | 合計 / 最大 / 最小 |
| `sorted(a)` | 小さい順に並べたリスト |
| `x in a` | x が a に含まれていれば True |

````python
scores = [72, 85, 90, 64]
scores.append(78)
print(scores, len(scores))
print(sum(scores), max(scores), min(scores))
print(sorted(scores))
print(90 in scores)
````

### for 文でリストを処理する
`for 変数 in リスト:` と書くと、要素を 1 つずつ取り出して繰り返します。

````python
for f in ["りんご", "バナナ", "みかん"]:
    print(f"{f}が好きです")
````

# 2. 例題

**【例題】** 5 人のテストの点数 `[72, 85, 90, 64, 78]` について、平均点を求め、平均点以上の点数だけを表示しなさい。

````python
scores = [72, 85, 90, 64, 78]
average = sum(scores) / len(scores)
print(f"平均点: {average}")
print("平均点以上の点数:")
for s in scores:
    if s >= average:
        print(s)
````

**【解説】**
- `sum()` と `len()` で平均を出しています。人数が変わってもそのまま使えます。
- for でリストの要素を 1 つずつ取り出し、if で条件に合うものだけ表示しています。
