# 1. 基本事項

### input() — 入力を受け取る
`input("メッセージ")` を実行すると入力欄が出ます。入力して **Enter** を押すと、その内容が **文字列（str）** として返されます。

````python
name = input("名前を入力してください: ")
print(f"ようこそ、{name}さん！")
````

### 数値として使うには変換が必要
`input()` で受け取った値は文字列なので、そのまま計算すると思った結果になりません。

````python
a = input("数を入力: ")
print(a * 3)          # 文字列のくり返しになってしまう！
b = int(input("もう一度数を入力: "))
print(b * 3)          # 整数として計算される
````

### よくあるエラー
- `ValueError: invalid literal for int()` … 整数に変換できない文字（`abc` や `3.5`）を `int()` に渡した
- 小数を入力させたいときは `float()` を使う
- セルが実行中（入力待ち）のまま次のセルを実行しようとしても進みません。入力欄に入力してから次へ。

# 2. 例題

**【例題】** 2 つの整数を入力させ、和・差・積を表示しなさい。

````python
a = int(input("1つ目の整数: "))
b = int(input("2つ目の整数: "))
print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} × {b} = {a * b}")
````

**【解説】**
- `int(input(...))` のように、入力と変換を 1 行で書くのが定番です。
- **入力 → 処理 → 出力** はプログラムの基本の形です。
