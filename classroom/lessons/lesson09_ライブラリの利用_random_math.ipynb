# 1. 基本事項

### import
Python には便利な機能をまとめた **ライブラリ（モジュール）** が用意されています。`import ライブラリ名` で読み込み、`ライブラリ名.関数名()` で使います。

### random — 乱数
| 書き方 | 意味 |
|---|---|
| `random.randint(a, b)` | a 以上 b 以下の整数をランダムに 1 つ |
| `random.choice(リスト)` | リストからランダムに 1 つ |
| `random.shuffle(リスト)` | リストの順番をランダムに並べかえる |

何度か実行して結果が変わることを確かめよう。

````python
import random

print(random.randint(1, 6))
print(random.choice(["グー", "チョキ", "パー"]))
cards = [1, 2, 3, 4, 5]
random.shuffle(cards)
print(cards)
````

### math — 数学
`math.sqrt()`（平方根）、`math.pi`（円周率）、`math.floor()`（切り捨て）、`math.ceil()`（切り上げ）など。

````python
import math

print(math.sqrt(2))
print(math.pi)
print(math.floor(3.7), math.ceil(3.2))
````

# 2. 例題

**【例題】** サイコロを 600 回振り、それぞれの目が出た回数を数えて表示しなさい。

````python
import random

counts = [0, 0, 0, 0, 0, 0]   # 1〜6の目の回数
for i in range(600):
    dice = random.randint(1, 6)
    counts[dice - 1] += 1

for i in range(6):
    print(f"{i + 1}の目: {counts[i]}回")
````

**【解説】**
- リストの添字は 0 から始まるので、`dice - 1` 番目を増やしています。
- 回数を 6000 回、60000 回…と増やすと、どの目も 1/6 に近づいていきます（大数の法則）。これは「シミュレーション」の基本です。
