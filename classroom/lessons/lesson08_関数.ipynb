# 1. 基本事項

### 関数とは
よく使う処理に名前をつけてまとめたものが **関数** です。`print()` や `len()` も関数です。自分で作るには `def` を使います。
```
def 関数名(引数):
    処理
    return 戻り値
```

````python
def greet(name):
    print(f"こんにちは、{name}さん！")

greet("太郎")
greet("花子")
````

### 引数と戻り値
- **引数**：関数に渡す値（カッコの中）
- **戻り値**：`return` で関数から返す値。返された値は変数に入れたり計算に使ったりできる

````python
def add(a, b):
    return a + b

x = add(3, 5)
print(x)
print(add(10, 20) * 2)
````

### return と print の違い
`print` は画面に表示するだけ。`return` は値を呼び出し元に返します。計算結果を後で使いたいときは `return` を使います。

````python
def square_print(n):
    print(n * n)

def square_return(n):
    return n * n

a = square_print(4)    # 16 と表示されるが…
b = square_return(4)
print(a, b)            # a は None（何も返していない）
````

# 2. 例題

**【例題】** 底辺と高さを受け取って三角形の面積を返す関数 `triangle_area` を作り、いくつかの三角形の面積を表示しなさい。

````python
def triangle_area(base, height):
    return base * height / 2

print(triangle_area(6, 4))
print(triangle_area(10, 3))
total = triangle_area(6, 4) + triangle_area(10, 3)
print("2つの合計:", total)
````

**【解説】**
- 一度関数を作れば、何度でも呼び出して使えます。
- `return` で値を返しているので、合計の計算にも使えます。
