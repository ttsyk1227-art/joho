# 1. 基本事項

### print() — 画面に表示する
`print()` のカッコの中に書いたものが表示されます。文字は `"` （ダブルクォーテーション）または `'` で囲みます。

````python
print("こんにちは")
print(123)
print("答えは", 10)   # カンマで区切ると、間にスペースを入れて並べて表示
````

### 算術演算子
| 記号 | 意味 | 例 | 結果 |
|---|---|---|---|
| `+` | 足し算 | `7 + 3` | 10 |
| `-` | 引き算 | `7 - 3` | 4 |
| `*` | かけ算 | `7 * 3` | 21 |
| `/` | 割り算 | `7 / 2` | 3.5 |
| `//` | 割り算の商（整数） | `7 // 2` | 3 |
| `%` | 割り算の余り | `7 % 2` | 1 |
| `**` | べき乗 | `2 ** 3` | 8 |

計算の順番は数学と同じ（`*` `/` が先）。`( )` で先に計算させられます。

````python
print(7 + 3)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 3)
print(2 + 3 * 4)
print((2 + 3) * 4)
````

### コメント
`#` から行末まではコメントです。Python は無視するので、メモや説明を書くのに使います。

# 2. 例題

**【例題】** 1 個 120 円のりんごを 5 個と、1 個 80 円のみかんを 3 個買いました。合計金額を表示しなさい。

````python
# りんごとみかんの合計金額
print("合計金額:", 120 * 5 + 80 * 3, "円")
````

**【解説】**
- `120 * 5` と `80 * 3` が先に計算され、その後に足し算されます。
- 文字（`"合計金額:"`）と計算結果を `,` で区切って並べて表示しています。
