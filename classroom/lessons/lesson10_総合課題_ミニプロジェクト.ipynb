# 2. 例題

**【例題】** じゃんけんゲーム：プレイヤーの手を入力させ、コンピュータの手をランダムに決めて勝敗を表示する。これを 3 回行い、最後に勝った回数を表示する。

````python
import random

hands = ["グー", "チョキ", "パー"]

def judge(player, computer):
    """勝ちなら"勝ち"、負けなら"負け"、同じなら"あいこ"を返す"""
    if player == computer:
        return "あいこ"
    elif (player - computer) % 3 == 2:
        return "勝ち"
    else:
        return "負け"

win = 0
for i in range(3):
    print(f"--- {i + 1}回戦 ---")
    p = int(input("0:グー 1:チョキ 2:パー  手を選んでください: "))
    c = random.randint(0, 2)
    result = judge(p, c)
    print(f"あなた: {hands[p]}  コンピュータ: {hands[c]}  → {result}")
    if result == "勝ち":
        win += 1
print(f"3回中{win}回勝ちました！")
````

**【解説】**
- 手を 0, 1, 2 の数で表し、リスト `hands` で名前に変換しています。
- 勝敗判定を関数 `judge` にまとめています。`(player - computer) % 3 == 2` は「グー(0)→チョキ(1)に勝つ」などの関係を 1 つの式で表したものです。自分で確かめてみよう。
- これまで学んだ **入力・条件分岐・繰り返し・リスト・関数・ライブラリ** をすべて使っています。
