# 1. 基本事項

### 比較演算子
| 記号 | 意味 |
|---|---|
| `==` | 等しい |
| `!=` | 等しくない |
| `>` / `<` | より大きい / より小さい |
| `>=` / `<=` | 以上 / 以下 |

比較の結果は `True`（正しい）か `False`（正しくない）になります。`=`（代入）と `==`（比較）の違いに注意！

````python
print(5 > 3)
print(5 == 3)
print(5 != 3)
````

### if / elif / else
```
if 条件1:
    条件1が True のときの処理
elif 条件2:
    条件1が False で、条件2が True のときの処理
else:
    どれも False のときの処理
```
- 条件の後ろに **`:`（コロン）** を忘れない
- 中の処理は **字下げ（インデント）** する。字下げされている部分が「その条件のときに実行される範囲」

````python
temp = 28
if temp >= 30:
    print("真夏日です")
elif temp >= 25:
    print("夏日です")
else:
    print("過ごしやすい気温です")
print("（この行は必ず実行される）")
````

### and / or / not
- `A and B` … A も B も True のとき True
- `A or B` … A か B のどちらかが True なら True
- `not A` … True と False を反転

````python
age = 16
if age >= 13 and age <= 18:
    print("中高生です")
````

# 2. 例題

**【例題】** テストの点数を入力させ、80点以上なら「A」、60点以上なら「B」、それ以外は「C」と表示しなさい。ただし 0〜100 以外が入力されたら「入力エラー」と表示する。

````python
score = int(input("点数: "))
if score < 0 or score > 100:
    print("入力エラー")
elif score >= 80:
    print("A")
elif score >= 60:
    print("B")
else:
    print("C")
````

**【解説】**
- 条件は **上から順に** 調べられ、最初に True になったところだけ実行されます。
- そのため、範囲外のチェックを先に書いています。80点以上の判定より後に 60点以上を書く順番も大切です。
