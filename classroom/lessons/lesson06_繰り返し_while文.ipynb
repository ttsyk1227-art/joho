# 1. 基本事項

### while 文
```
while 条件:
    条件が True の間くり返す処理
```
回数が決まっていない繰り返しに使います。条件がいつか False になるように、中で変数を変化させるのを忘れずに。

````python
n = 1
while n <= 5:
    print(n)
    n += 1
print("終わり")
````

### break — 繰り返しを抜ける
`break` を実行すると、その場で繰り返しを終了します。`while True:`（ずっと繰り返す）と組み合わせてよく使います。

````python
while True:
    word = input("文字を入力（endで終了）: ")
    if word == "end":
        break
    print(f"「{word}」が入力されました")
print("終了しました")
````

### ⚠️ 無限ループに注意
条件がずっと True のままだと止まりません。セル左の **■（停止）ボタン**、または「ランタイム」→「実行を中断」で止められます。

# 2. 例題

**【例題】** 貯金箱に毎日 500 円ずつ貯金します。10000 円以上になるのは何日目か求めなさい。

````python
money = 0
day = 0
while money < 10000:
    day += 1
    money += 500
print(f"{day}日目に{money}円になります。")
````

**【解説】**
- 「何回繰り返すか」が最初はわからないので while を使います。
- `money < 10000` の間だけ繰り返し、10000 以上になったら抜けます。
