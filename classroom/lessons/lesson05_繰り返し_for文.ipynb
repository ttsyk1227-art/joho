# 1. 基本事項

### for 文と range()
```
for 変数 in range(回数):
    くり返す処理
```
`range()` は数の並びを作ります。
| 書き方 | 作られる数 |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5（6 は含まない） |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8（2 ずつ増える） |

````python
for i in range(5):
    print(i)
print("---")
for i in range(1, 11, 3):
    print(i)
````

### 合計を求める（累積）
合計用の変数を **0 で初期化** してから、繰り返しの中で足していきます。`total += i` は `total = total + i` の省略形です。

````python
total = 0
for i in range(1, 6):
    total += i
    print(f"{i}を足して合計{total}")
print("合計:", total)
````

### 二重ループ
for の中に for を書くと、外側 1 回ごとに内側が全部繰り返されます。`print(..., end="")` を使うと改行せずに表示できます。

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end=" ")
    print()   # 改行
````

# 2. 例題

**【例題】** 1 から 100 までの整数のうち、3 の倍数の合計を求めなさい。

````python
total = 0
for i in range(1, 101):
    if i % 3 == 0:
        total += i
print("3の倍数の合計:", total)
````

**【解説】**
- `range(1, 101)` で 1〜100 を順に取り出します。
- 繰り返しの中に if 文を入れて、条件に合うものだけを足しています（for と if の組み合わせ）。
- `range(3, 101, 3)` を使えば if なしでも書けます。試してみよう。
