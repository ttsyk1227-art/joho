```python
import matplotlib.pyplot as plt

x = list(range(1, 11))
y = [v ** 2 for v in x]

plt.plot(x, y, marker="o")
plt.title("y = x^2")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
